# Chemical simulation interface
This notebook provides an interface for a toy model of a chemical reaction in the gas phase.

```python
from collider import Simulation, Wall #have to add 'Wall' so it is defined 
from ipycanvas import Canvas
```

## Initialize and run the simulation
Initialize the Simulation class with the desired parameters and run the simulation loop.

```python
'''Adding a solid wall in the middle of the collider'''

simulation = Simulation(Canvas, 800, 600, 100, temperature=100)

mid = simulation.width / 2
simulation.walls.append(Wall(mid, 0, mid, simulation.height))  # vertical wall in the centre

simulation.canvas #has to be last line of the cell, or else Jupiter won't display it 
#alternatively: display(simulation.canvas) anywhere in cell
```

```python
simulation.run()
```
