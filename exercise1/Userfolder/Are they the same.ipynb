```python
%matplotlib inline

import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
from pathlib import Path
```

# Do we think the same thing is happening in all of these countries?

This is a hypothetical dataset representing pesticide residues above a safety threshold vs. percentage of organic farms in different countries. How can we tell if trends are consistent across countries?

```python
for file in Path('.').glob("*.csv"):
    df = pd.read_csv(file)
    x, y = df['x'], df['y']
    x_mean = np.mean(x)
    y_mean = np.mean(y)
    coeff = np.polyfit(x, y, deg=1)
    corr = np.corrcoef(x, y)[0, 1]
    
    print(f"x_mean={x_mean:.3f}, y_mean={y_mean:.3f}, linear regression fr(x)={coeff[1]:.1f} {coeff[0]:.2f} x, r = {corr:.3f} ")
```

```python
for file in Path('.').glob("*.csv"):
    df = pd.read_csv(file)
    x, y = df['x'], df['y']
    plt.scatter(x,y)
    # 2. Calculate the linear regression line (degree 1)
    # slope = coeff[0], intercept = coeff[1]
    slope, intercept = np.polyfit(x, y, deg=1)
    
    # 3. Create x values for the line and calculate corresponding y values
    x_line = np.linspace(x.min(), x.max(), 100)
    y_line = slope * x_line + intercept
    
    # 4. Plot the regression line
    plt.plot(x_line, y_line, color='red', linestyle='--', label=f'Fit: y={slope:.2f}x+{intercept:.2f}')
    plt.show()
```
